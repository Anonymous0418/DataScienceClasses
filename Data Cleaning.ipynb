## Data Cleaning

```python
'''
1. Handle Missing / Null Values
    a. Fill Null Values
        -> .fillna()
        - Continuous Data
        i. Mean
        ii. Median
            
        -> Skewness -> .skew()
            -> 0 - 0.5 -> Fill Mean Value
            -> < 0 (Negative Skewness) or > 0.5 (Positive Skewness) -> Fill Median Value
            
        -> Calculate Mean and Median value of a column.
            -> If both mean and median value are equal or almost equal then fill mean value.
            -> If both mean and median value are not equal has vast difference in values then fill median value.
            
        - Discrete / Categorical
        iii. Mode
            
        - Continuous / Discrete / Categorical
        iv. Custom Record
        
        
    b. Drop Null Values
        -> .dropna()

2. Remove unwanted data
    -> Remove Whitespace
        -> .strip(), .lstrip(), .rstrip()
        
    -> Remove Regular Expressions (regex) if needed / Remove unwanted texts
        -> .replace()

3. Change Datatypes
    .astype()

4. Handle Duplicate Records
    -> Find duplicate records
        -> .duplicated()
        
    -> To remove duplicate records
        -> .drop_duplicates()
'''
```

## Handle Duplicate Records

```python
duplicated_data = df[df.duplicated(subset=['EMPLOYEE_ID'], keep=False)]
duplicated_data
```

### Remove duplicate records

```python
df.drop_duplicates(subset='EMPLOYEE_ID', keep='last', inplace=True)
```

## Fill commission value

```python
commission_mean=round(df['COMMISSION_PCT'].mean(), 2)
commission_median = df['COMMISSION_PCT'].median()
commission_skew = df['COMMISSION_PCT'].skew()
```
