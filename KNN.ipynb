```python
def convertMarks(marks):
    if marks >=15:
        return 'A'
    elif marks >=11:
        return 'B'
    elif marks >=7:
        return 'C'
    else:
        return 'Fail'
```

### Feature Encoding

```python
# Label Encoder
le = LabelEncoder()

def encodeData(dataframe, *args):
    for col in args:
        dataframe[f'{col}_id'] = le.fit_transform(dataframe[col])
    print('Columns Encoded.')
    
    
encodeData(df, 'schoolsup', 'famsup', 'nursery', 'internet')
```

```python
df['health'].unique()
```

```python
medu_id = {
    'none': 0,
    'primary education (4th grade)': 1,
    '5th to 9th grade': 2,
    'secondary education': 3,
    'higher education': 4
}
df['medu_id'] = df['Medu'].map(medu_id)
```

```python
# Manual Encoding
def menualEncode(dataframe, col, *args):
    encode_val = {}
    for idx, val in enumerate(args):
        encode_val[val] = idx

    dataframe[f'{col}_id'] = dataframe[col].map(encode_val)
```

```python
menualEncode(df, 'Fedu', 'none', 'primary education (4th grade)', '5th to 9th grade','secondary education',
        'higher education')
```
